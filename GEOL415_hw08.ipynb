```python
# importing libraries
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import seaborn as sns
from sklearn import linear_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
waves = pd.read_csv('/Users/rubyestrickland/Desktop/EnvDataSci/HW08/Waves_2023.txt', sep=r'\s+')
```

```python
waves.describe()
```

```python
waves= waves.drop(['WDIR','WSPD','GST','PRES','ATMP','DEWP','VIS','TIDE'], axis=1)
waves.describe()
```

```python
# WVHT = Wave Height (m)
# DPD = Dominant wave period (seconds)
# APD = Average wave period (seconds)
# MWD = The direction from which the waves at the dominant period (DPD) are coming
# WTMP = Sea surface temperature (Celsius)

waves = waves[waves['WVHT'] < 99]
waves = waves[waves['DPD'] < 99]
waves = waves[waves['APD'] < 99]
waves = waves[waves['MWD'] < 999]
```

```python
wavesML = waves.drop(['YY','MM','DD','hh','mm'],axis = 1)
```

```python
wavesML
```

```python
df_X = wavesML.drop(['WVHT'],axis=1).to_numpy()
df_y = wavesML['WVHT'].to_numpy()

X_train, X_test, y_train, y_test = train_test_split(df_X, df_y, test_size=0.10)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.10)
```

```python
from sklearn.ensemble import RandomForestRegressor
regRF = RandomForestRegressor(n_estimators=100)
regRF = regRF.fit(X_train, y_train)
yPred = regRF.predict(X_val)
mean_squared_error(yPred,y_val)
```

```python
regRF = RandomForestRegressor(n_estimators=25)
regRF = regRF.fit(X_train, y_train)
yPred = regRF.predict(X_val)
val_mse = mean_squared_error(yPred,y_val)
print(val_mse)
```

```python
yPred_test = regRF.predict(X_test)
test_mse = mean_squared_error(yPred_test, y_test)
print(test_mse)
```

```python
yPred_train = regRF.predict(X_train)
train_mse = mean_squared_error(yPred_train, y_train)
print(train_mse)
```

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df_X, df_y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
model = Ridge(alpha=1.0)
model.fit(X_train_scaled, y_train)
```

```python
y_pred_train = model.predict(X_train_scaled)
y_pred_test = model.predict(X_test_scaled)
train_mse = mean_squared_error(y_train, y_pred_train)
test_mse = mean_squared_error(y_test, y_pred_test)
```

```python
print(test_mse)
```

```python
X_val_scaled = scaler.transform(X_val)
y_pred_val = model.predict(X_val_scaled)
val_mse = mean_squared_error(y_val, y_pred_val)
print(val_mse)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df_X, df_y, test_size=0.10)
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.10)
```

```python
regRF = RandomForestRegressor(n_estimators=25)
regRF = regRF.fit(X_train, y_train)
yPred = regRF.predict(X_val)
val_mse = mean_squared_error(yPred,y_val)
print(val_mse)
```

```python
yPred_test = regRF.predict(X_test)
test_mse = mean_squared_error(yPred_test, y_test)
print(test_mse)
```

```python
yPred_train = regRF.predict(X_train)
train_mse = mean_squared_error(yPred_train, y_train)
print(train_mse)
```

```python
tree_counts = [2, 10, 25, 50, 100, 250, 500, 1000]
results_data = []

for n_trees in tree_counts:
    rf_model = RandomForestRegressor(n_estimators=n_trees, random_state=42, n_jobs=-1)
    rf_model.fit(X_train, y_train)
    
    y_pred_train = rf_model.predict(X_train)
    y_pred_val = rf_model.predict(X_val)
    y_pred_test = rf_model.predict(X_test)
    
    train_mse = mean_squared_error(y_train, y_pred_train)
    val_mse = mean_squared_error(y_val, y_pred_val)
    test_mse = mean_squared_error(y_test, y_pred_test)
    
    results_data.append({
        "Number of Trees": n_trees,
        "Training MSE": round(train_mse, 4),
        "Validation MSE": round(val_mse, 4),
        "Test MSE": round(test_mse, 4)
    })

df_results = pd.DataFrame(results_data)
print(df_results.to_string(index=False))
```
